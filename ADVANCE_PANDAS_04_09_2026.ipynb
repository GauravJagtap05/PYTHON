```python
# Q2 Write a python program to find out and display horizontal bar chart the alcohol consumption details in the year '1987' where WHO region is 'Western Pacific'
# and country is 'VietNam' from the dataset.
```

```python


df_Q2 = df_Al[(df_Al['Year']==1987) & ((df_Al['WHO region']== 'Western Pacific') & (df_Al['Country']== 'Viet Nam'))]
df_Q2.dropna()
df_Q2
# df.plot.bar(x='Open', y='Close', alpha = 0.5)

```

```python
import matplotlib.pyplot as plt
```

```python

plt.bar("Beverages Type", 'Display Value', data = df_Q2, color ='Orange', label='Year 1989')
plt.xlabel('ALCHOL CONSUMPTION')
plt.ylabel('Western Pacific -Viet Nam ')
plt.show()
```

```python
# Q3 Write a Pandas program to find and display bar chart to out the alcohol consumption details in the year '1986' or '1989'
# where WHO region is 'Americas' or 'Europe' from the world alcohol consumption dataset.
# Use Different colors for each bar and attach text label above each bar.
```

```python

df_Q3 = df_Al[(df_Al['Year'] ==1986) | (df_Al['Year'] == 1989).where((df_Al["WHO region"] == 'Americas') | (df_Al["WHO region"] == 'Europe'))]
df_Q3.head()
```

```python
plt.bar("WHO region", 'Display Value', data = df_Q3, color ='green', label='1989')
plt.xlabel('"WHO region"')
plt.ylabel('Alcohol Consumption')
plt.title('1989')
plt.show()
```

```python
# Q4
# Write a Pandas program to find out the 'WHO region, 'Country', 'Beverage Types' in the year '1986' or '1989'
# where WHO region is 'Americas' or 'Europe' from the world alcohol consumption dataset.
```

```python
df_Q4 = df_Al[(df_Al['Year'] ==1986) | (df_Al['Year'] == 1989).where(df_Al["WHO region"] == 'Americas') | (df_Al["WHO region"] == 'Europe')][['WHO region','Country','Beverage Types']].dropna()
df_Q4.head()
```

```python
# Q5
# Write a Pandas program to find out the records where consumption of beverages per person average >=5
# and Beverage Types is Beer from world alcohol consumption dataset.
```

```python
# Q6
#Write a Pandas program to find out and display Pie chart of the records where consumption of beverages per person average >=4
# and Beverage Types is Beer, Wine, Spirits from world alcohol consumption dataset.
```

```python
r1 = df_Al[(df_Al['Display Value'] >=4) & (df_Al['Beverage Types'] == 'Beer')]
r2 = df_Al[(df_Al['Display Value'] >=4) & (df_Al['Beverage Types'] == 'Wine')]
r3 = df_Al[(df_Al['Display Value'] >=4) & (df_Al['Beverage Types'] == 'Spirits')]
```

```python

plt.pie(r1,shadow =True, autopct ='%.1f%%')
plt.show()
plt.pie(r2,shadow =True, autopct ='%.1f%%')
plt.show()
plt.pie(r3,shadow =True, autopct ='%.1f%%')
plt.show()
```
