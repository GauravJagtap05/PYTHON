```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'Player_name':['Virat Kolhi', 'Rohit Sharma', 'Rahul Dravid'], 'Age':[20,30,40]})
df1
```

```python
df2 = pd.DataFrame({'Player_name':['Virat Kolhi', 'Rohit Sharma', 'ishan sharma','Hardik Pandya'], 'Salary':[23,45,56,50]})
df2
```

```python
# inner join

pd.merge(df1,df2, on='Player_name',how='inner')
```

```python
# left join
pd.merge(df1,df2, on='Player_name',how='left')


```

```python
# right join
pd.merge(df1,df2, on='Player_name',how='right')
```

```python
# outer join
pd.merge(df1,df2, on='Player_name',how='outer')
```

```python
# how to join when column names does not match ??

df3 = pd.DataFrame({'Sports_person':['Virat Kolhi', 'Rohit Sharma', 'ishan sharma','Hardik Pandya'], 'Salary':[23,45,56,50]})
df3

```

```python
# merge
pd.merge(df1,df3, left_on = 'Player_name', right_on='Sports_person', how='inner')
```

```python
# merge left
pd.merge(df1,df3, left_on = 'Player_name', right_on='Sports_person', how='left')
```

```python
# CONCAT -----------------------------------------------
# to join the tables side by side
```

```python
# concat always pas the data frames in list format

pd.concat([df1,df3])
```

```python
# concat always pas the data frames in list format

pd.concat([df1,df2,df3])
```

```python
# on the above dataframe we want to change the index

pd.concat([df1,df2,df3], ignore_index='True')
```

```python
# Concat column wise

pd.concat([df1,df2], axis =1)
```
