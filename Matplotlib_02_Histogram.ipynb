```python
import matplotlib.pyplot as plt
```

```python
# HISTOGRAM -- COUNT THE ELEMENTS IN A RANGE

emp_ages = [33,33,21,24,56,50,445,45,50,24,45,55,55,45,33,45]
bins = [0,10,20,30,40,50,60]
```

```python
# use hist() function for histogram
# histtype can be =  bar, barstacked, step, stepfilled
plt.hist(emp_ages, bins, histtype = 'bar', rwidth= 0.9, color='white')
plt.xlabel(' employees age')
plt.ylabel(' no of employees')
plt.title(' AVD  GROUP, PUNE , BATCH 2026')
plt.show()
```
