```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# working with dataset IRIS


df = sns.load_dataset('iris')
df.head(10)
```

```python
len(df)
```

```python
df.info()
```

```python
# PAIR PLOT -- IT REPRESENT RELATIONSHIP BETWEEN ALL COLUMNS
# markers -- round shapes dots changes to another shapes / 'o' -circle , 'd' - dimond, 's' - square, * - star,
# pallet colours :- BrBG, Blues, husl, coolwarm, cubehelix

sns.pairplot(df, hue = 'species', palette='BrBG', markers = ['^','d','*'])
```

```python
# RELATION B/W ONLY 2 COLUMNS

sns.pairplot(df, vars= ['sepal_length', 'sepal_width'])
```

```python
# HEAT MAP

df1 = df.iloc[0:10, 0:4]
df1
```

```python
sns.heatmap(df1, annot = True, cmap = 'coolwarm')
```

```python
# CATPLOT
sns.catplot(x ='species', y = 'sepal_length', data = df, kind= "violin")
```
