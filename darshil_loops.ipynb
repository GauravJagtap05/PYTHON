```python
# list comprehensions:

names = ['ravi','raj','sandy','arav']
hello = ['Hello ' + name for name in names]
print(hello)
```

```python
numbers = [1,23,45,3,4,567,6,5,323,54,67,88,99,90]
even = [ num for num in numbers if num % 2 ==0 ]
print(even)

odd = [num for num in numbers if num % 2 != 0]
print(odd)

odd_square = [(num, num*num) for num in numbers if num % 2 == 1]
print(odd_square)
```
