```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# importing the data
df = pd.read_csv("D:/AVD/PYTHON/datasets-20260827T080306Z-1-001/datasets/datasets1/datasets1/score.csv")
df
```

```python
# LINE PLOT

x = df['marks']
y= df['rank']
```

```python
# visulization using line graph

plt.plot(x,y, color = 'green')
plt.xlabel('Marks', color ='blue')
plt.ylabel('rank', color ='red')
plt.title('MPSC Exam - 2026, Maharashtra ', color = 'Green')
plt.show()
```

```python
# creating a SCATTER PLOT

x = df['marks']
y= df['rank']
```

```python
# visulization using scatter plot

plt.scatter(x,y, color = 'yellow')
plt.plot(x,y, color = 'green')
plt.xlabel('Marks', color ='blue')
plt.ylabel('rank', color ='red')
plt.title('MPSC Exam - 2026, Maharashtra ', color = 'Green')
# plt.legend()
plt.show()
```

```python
# BAR PLOT
x = df['marks']
y= df['rank']
```

```python
plt.bar(x,y, color= 'cyan', label = 'MY BAR PLOT')
plt.xlabel('marks', color= 'red')
plt.ylabel('rank', color= 'red')
plt.title('JEE EXAM -2026 ', color ='orange')
plt.xlim(500,1000)
plt.ylim(1100,2100)
plt.show()
```

```python
# CREATING THE PIE CHART

x = df['marks']

# HERE X IS A DATAFRAME
# x[x>950]
```

```python
s1 = len(x[x>950])
s2 = len(x[(x>850) & (x<= 950)])
s3 = len(x[x<=850])
print(s1,s2,s3)
```

```python
slices = [s1,s2,s3]
cols =['pink','white','brown']
lbls =['above 950', '850 to 950', 'less than 850']
plt.pie(slices, colors = cols, labels = lbls , startangle = 270 , shadow = True,
        autopct = '%.1f%%', explode= (0,0.2,0 )) # we can change the angle of the pie chart using startangle = 180
plt.show()
```

```python
# SUB PLOTS (you can accomodate more than one plot side by side on a single plage )
# subplot --- figure object

x =df['marks']
y =df['rank']
z= df['admit']
```

```python
# subplot --- figure object

fig = plt.figure(figsize = (10,10))
g1 = fig.add_subplot(2,2,1, facecolor = 'white')
g1.plot(x,y)

g1.set_title('marks vs rank')
g2 = fig.add_subplot(2,2,2, facecolor = 'lightgreen')
g2.plot(x,z)
g2.set_title(" marks vs admit")

g3 = fig.add_subplot(2,2,3, facecolor = 'lightgreen')
high = max(x) # max marks
low = min(x) # min of marks
n = len(z[z==1])  # no .of students admitted
g3.pie([high,low,n], colors = ['pink', 'maroon','lightgreen'], explode= (0,0.1,0 ), startangle = 90,
       autopct = '%.1f%%',labels = lbls )
g3.set_title(" marks vs admission")
```
