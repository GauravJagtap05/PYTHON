```python
# seaborn is an advance package for data visulization

import pandas as pd
import seaborn
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("D:/AVD/PYTHON/datasets-20260827T080306Z-1-001/datasets/datasets1/datasets1/mpg.csv")
df
```

```python
mgp_df = sns.load_dataset('mpg')
mgp_df.head()
```

```python
# visualize the dat in he form of graph and plots

mgp_df['cylinders'].unique()
```

```python
mgp_df['horsepower'].unique()
```

```python
# distribution plot
tpl = tuple(mgp_df['cylinders'])
sns.displot(tpl, kde = True, bins=range(3,10,1))
plt.xlabel('no of cylinders')
plt.ylabel('no of cars')
plt.show()
```

```python
# count plot
sns.countplot(x ='cylinders', data=mgp_df, palette='dark:green', hue = 'origin')


```

```python
# BOX PLOT

sns.boxplot(x ='origin', y = 'acceleration', data = mgp_df, palette='dark:pink', hue = 'origin')
```

```python
# joint plot

sns.jointplot(x ='horsepower', y = 'acceleration', data = mgp_df, hue = 'origin')
```

```python
# Scatter Plot

sns.scatterplot(x ='horsepower', y = 'acceleration', data = mgp_df, hue = 'origin', alpha = 0.5) # alpha - transperent
```

```python
# SUBPLOTS

plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
sns.scatterplot(x ='horsepower', y = 'acceleration', data = mgp_df, hue = 'origin') # hue --changing the color


plt.subplot(2,2,2)
sns.boxplot(x ='origin', y = 'acceleration', data = mgp_df, hue = 'origin') # hue --changing the color

plt.subplot(2,2,3)
sns.scatterplot(x ='horsepower', y = 'origin', data = mgp_df, hue = 'origin') # hue --changing the color

plt.subplot(2,2,4)
sns.boxplot(x ='origin', y = 'acceleration', data = mgp_df ,hue = 'origin')

plt.savefig('D:/AVD/PYTHON/seaborn_my_fig.jpeg')

```

```python
#LMPLOT

sns.lmplot(x ='horsepower', y = 'acceleration', data = mgp_df, hue = 'origin',palette='dark:blue') # hue --changing the color
```
