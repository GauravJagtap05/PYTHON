```python
# number that divisible with 1, itself

n = int(input("enter the number ; "))
count = 0

for i in range(1, n+1):
    if n % i ==0:
        count +=1
if count == 2:
    print(" its a prime number")
else:
    print(" its not a prime number ")
```

```python
# Write a Python program that takes the number of units consumed and calculates
# the electricity bill according to these slabs:
# First 100 units  → 100 × ₹5
# Next 100 units   → 100 × ₹7
# Next 50 units    → 50 × ₹10

n = int(input('enter a number units: '))
if n <= 100:
    bill = n * 5
    print(f'the electricity bill for this month is {bill}')
elif n > 100 and n <= 200:
    bill = n * 7
    print(f'the electricity bill for this month is {bill}')
else:
    bill = n * 10
    print(f'the electricity bill for this month is {bill}')
```

```python
# gussing game

import random

jackpot = random.randint(1,100)

guess_num = int(input('guess the number'))
counter = 1

while guess_num == jackpot:
    if guess_num <jackpot:
        print(' wrong number guess higher')
    else:
        print(' wrong number guess lower')
    guess_num = int(input('guess the number'))
    counter +=1

else:
    print('correct guess')
    print('attempts', counter)

# o/p --  wrong number guess higher
# guess_num < jackpot




```
