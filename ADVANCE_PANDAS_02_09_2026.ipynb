```python
import pandas as pd
import numpy as np
```

```python
# WINDOW FUNCTIONS -----------------------
# 1. CUMULATIVE WINDOW
# 2. EXPANDING WINDOW
# 3. ROLLING WINDOW
```

```python
df = pd.DataFrame({'a':[1,2,3,4,5,6,7,8],
                   'b':[10, np.nan, np.nan, 20, np.nan, np.nan, 30,30]})
```

```python
df
# if you want to create an empty value give np.nan -- value
```

```python
# creating a rolling window
# rolling sum of col'a' from window size =2
```

```python
obj = df['a'].rolling(window=2) # -- on df of column[a] we have created the rolling window = 2
#obj.sum() --- doing sum aggreagtin on that column a
df['a_roll_sum'] = obj.sum()  # --- a_roll_sum is the column name is created and sum of column
# a will be stored on that column
df
```

```python
# rolling count on column 'b'

obj = df['b'].rolling(window=3, min_periods =3)
df['b_roll_count'] = obj.count()
```

```python
df[['b','b_roll_count']]
```

```python

# CUMULATIVE ROLLING SUM
```

```python
# CUMULATIVE SUM ON THE COL 'a'

df['a_cumsum'] = df['a'].cumsum()

df
```

```python
# EXPANDING WINDOW
# EXPANDING FUNCTIONS ARE LIKE CUMULATIVE FUNCTIONS ONLY

obj = df['a'].expanding(min_periods=1)
df['expanding_sum'] = obj.sum()
df
```

```python
obj = df['a'].expanding(min_periods=2)
df['expanding_sum_01'] = obj.sum()
df
```

```python
# find mean of an expanding window

obj = df['a'].expanding(min_periods=1)
df['a_expanding_mean'] = obj.mean()
df
```
