```python
mydict ={'col1':[1,2,3,4], 'col2':[5,6,7,8], 'col3':['one','two','three','four']}
# in this dict the keys will become columns of the df
import pandas as pd

df = pd.DataFrame(mydict)
df

```

```python
# TO RENAME THE COLUMNS NAMES
# using rename method()
# always give the column names in dictonary
df1 = df.rename(columns ={'id':'EmpId', 'names':'Names'})
df1
```

```python
# TO RENAME THE COLUMNS NAMES
# using rename method()
# always give the column names in dictonary
# if we don't want to modify the data frame in same data frame we want to modify the data then use
# inplace =True

df.rename(columns ={'id':'EmpId', 'names':'Names'}, inplace =True)
df
```
